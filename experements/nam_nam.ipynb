```python
from qiskit import QuantumCircuit
import os
from os.path import join, isfile
import csv
```

```python
def get_circuit_info(fn):
    circ = QuantumCircuit.from_qasm_file(fn)
    count_dict = circ.count_ops()

    return (
        sum([count_dict[k] for k in count_dict]),
        count_dict["cx"],
        circ.depth(),
    )


circuit_order = [
    "tof_3",
    "barenco_tof_3",
    "mod5_4",
    "tof_4",
    "barenco_tof_4",
    "tof_5",
    "mod_mult_55",
    "vbe_adder_3",
    "barenco_tof_5",
    "csla_mux_3",
    "rc_adder_6",
    "gf2^E4_mult",
    "tof_10",
    "mod_red_21",
    "hwb6",
    "gf2^E5_mult",
    "csum_mux_9",
    "barenco_tof_10",
    "qcla_com_7",
    "ham15-low",
    "gf2^E6_mult",
    "qcla_adder_10",
    "gf2^E7_mult",
    "gf2^E8_mult",
    "qcla_mod_7",
    "adder_8",
]

path = "./output/nam_nam/"
fns = [
    fn for fn in os.listdir(path) if isfile(join(path, fn)) and fn[-4:] == "qasm"
]
result_dict = {}
for i, fn in enumerate(fns):
    if fn == "csum_mux_9_before_corrected.qasm":
        continue

    gate_cnt, cnot_cnt, depth = get_circuit_info(path + fn)
    result_dict[fn] = [gate_cnt, cnot_cnt, depth]


csv_file = "./metricks/nam_nam.csv"

with open(csv_file, mode="w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["filename", "total_gates", "cx_gates", "depth"])

    for fn in circuit_order:
        if fn + "_after_heavy.qasm" in result_dict:
            writer.writerow([fn + "_after_heavy.qasm"] + result_dict[fn + "_after_heavy.qasm"])
```
