```python
from qiskit import QuantumCircuit
from qiskit.circuit import Parameter
from qiskit.converters import circuit_to_dag, dag_to_circuit


qc = QuantumCircuit(4)

theta = Parameter('π/4')

qc.rz(theta, 0)
qc.x(0)

qc.rz(theta, 1)
qc.h(2)
qc.cx(1, 2)

qc.x(3)

qc.h(1)
qc.cx(2, 3)
qc.cx(1, 2)

qc.h(3)
qc.rz(theta, 3)

qc.h(2)

qc.cx(0, 1)


# fig = qc.draw(output='mpl', style='bw')
# fig.savefig("./img/1.png")


dag = circuit_to_dag(qc)

dag.draw(filename="../img/dag_1.png")

```

```python
from qiskit import QuantumCircuit
from qiskit.converters import circuit_to_dag, dag_to_circuit

qc = QuantumCircuit(4)
qc.h(1)
qc.h(2)
qc.cx(2, 3)
qc.h(3)
qc.cx(1, 2)
qc.h(2)
qc.cx(0, 1)
qc.h(0)
qc.h(1)




# fig = qc.draw(output='mpl', style='bw')
# fig.savefig("./img/common_5_start.png",transparent=True)

dag = circuit_to_dag(qc)

dag.draw(filename="../img/common_5_dag_1.png")

qc = QuantumCircuit(4)
qc.h(1)
qc.h(2)
qc.cx(2, 3)
qc.h(3)
qc.cx(1, 2)
qc.h(2)
qc.h(0)
qc.h(1)
qc.cx(1, 0)

# qc.draw(output='mpl', style='bw')
fig = qc.draw(output='mpl', style='bw')
fig.savefig("./img/common_5_1_start.png",transparent=True)

qc = QuantumCircuit(4)
qc.h(1)
qc.h(2)
qc.cx(2, 3)
qc.h(3)
qc.h(1)
qc.h(2)

qc.cx(2, 1)
qc.h(0)
qc.cx(1, 0)

# qc.draw(output='mpl', style='bw')
fig = qc.draw(output='mpl', style='bw')
fig.savefig("./img/common_5_2_start.png",transparent=True)

qc = QuantumCircuit(4)
qc.h(1)
qc.h(2)
qc.h(2)
qc.h(3)

qc.cx(3, 2)
qc.h(1)

qc.cx(2, 1)
qc.h(0)
qc.cx(1, 0)

# qc.draw(output='mpl', style='bw')
fig = qc.draw(output='mpl', style='bw')
fig.savefig("./img/common_5_3_start.png",transparent=True)

qc = QuantumCircuit(4)
qc.h(3)

qc.cx(3, 2)

qc.cx(2, 1)
qc.h(0)
qc.cx(1, 0)

# qc.draw(output='mpl', style='bw')
fig = qc.draw(output='mpl', style='bw')
fig.savefig("./img/common_5_4_finish.png",transparent=True)
```
